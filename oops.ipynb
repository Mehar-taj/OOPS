# OOPS-Object Oriented Programming System
OOPs is a programming approch where we organize our program using classes and objects.

## Use of OOPs
* Organize large program
* Reuse code
* Protect data
* Make program easier to maintain
* Represent real-world things in code

## 1. CLASS
* A class is a blueprint for creating objects. It contains attributes and methods that define the properties and behaviour of objects.

```python
class Student: # Class
    name = "Mehar" #  Attributes
    age = 21

    def study(): # Method
        print("Student is studing")
```

## 2. OBJECT
* An object is an instance of a class. It represents the real entity and can access the attributes and methods defined by the class.

```python
class Student: # Creating class
    name = "Mehar" # name -- attribute
    age = 21

student1 = Student() # Creating object

print(student1.name)
print(student1.age)
```

## 3. MULTIPLE OBJECTS
* One class can create multiple objects

```python
class Student():
    clg = "ALITS"

student1 = Student()
student2 = Student()
student3 = Student()

print(student1.clg)
print(student2.clg)
```

## 4. CONSTRUCTOR -- __init()__
* Constructor = The concept of creating Object.
* __init()__ = __init__() is a special method that automatically called when an object is created. It is used to initialize the object's attributes.


```python
class Student():

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name:{self.name}, Age:{self.age}")

Student1 = Student("Mehar", 21)
Student2 = Student("Divya", 22)

Student1.display()
Student2.display()
```

# 6. Attributes & Methods
* Attributes - Represents data/properties of an object.
* Methods - A function defined inside the class is called a method. It represents behaviour/action of an object.

```python
class students():

    def __init__(self,name,age):
        self.name = name # Attributes
        self.age = age

    def study(self): # Method
        print(self.name, self.age)

s1 = students('mehar',21)
s2 = students("saba",23)

print(s1.name)
print(s2.name)

s1.study()
s2.study()
```

# 8. Inheritance
Inheritance means one class can use the properties and mehoda of another class. It helps in code reuseability and reduces duplicate code.
## Why do we use it ?
* code reusability -- instead of writing the same code again, we can inherit it.
## Types of Inheritance
1. single       - one parent-->one child
2. Multiple     - Multiple parents --> one child
3. Multilevel   - Grandparent-->parent-->Child
4. Hierarchical - One parent-->multiple childern
5. Hybrid       - Combination of inheritance types


```python
class parent():

    def show(self):
        print("I am parent")

class child(parent):
    pass

ob1 = child()
ob1.show() # Even though show() is written inside parent class, the child can use it
```

```python
# Example using __init__()
class person():
    def __init__(self,name):
        self.name = name
    def display(self):
        print("Name:",self.name)

class student(person):
    pass

s1 = student("Mehar")
s2 = student("saba")

s1.display()
s2.display()
```

```python
# Single Inheritance
class animal:

    def eat(self):
        print("Eating")

class dog(animal):

    def bark(self):
        print("Barking")

d = dog()
d.eat()
d.bark()
```

```python
# Multiple Inheritance
class father():

    def father_prop(self):
        print("Father's property")

class mother():

    def mother_prop(self):
        print("Mother's property")

class child(father,mother):
    pass

c = child()
c.father_prop()
c.mother_prop()
```

```python
# Multilevel Inheritance
class grandparent:

    def house(self):
        print("Grandparent's house")

class parent(grandparent):

    def car(self):
        print("parent's car")

class child(parent):

    def bike(self):
        print("bike")

c = child()
c.house()
c.car()
c.bike() # Child can access methods from both grandparent and parent
```

### super() in Inheritance
super() is used to call the parent claass's method from the child class.<br>
or<br>
super() is used to access methods or the constructor of the parent classs from the child class.

```python
# Example
class parent:
    def show(self):
        print("Parent method")

class child(parent):
    def show(self):
        super().show()
        print("Child method")

c = child()
c.show()
```

```python
# super() with __init__()
class person():

    def __init__(self,name,roll_no):
        self.name = name
        self.roll_no = roll_no

class student(person):

    def __init__(self,name,roll_no,age):
        super().__init__(name,roll_no)
        self.age = age

c1 = student("mehar",35,21)
print(c1.name)
print(c1.roll_no)
print(c1.age)
```

# 9.Polymorphism
Polymorphism allows the same method or interface to have different behaviours depending on the object.<br>
**Polymorphism = One name,many forms**

## 3 imp forms of polymorphism
1. Method Overriding
2. Duck Typing
3. Operator Overloading

***Method Overriding*** occurs when a child class provides a different implememtation of a method that is already defined in it's parent class.

***Duck Typing*** means Python focuses on what an object can do rather than what type of object it is.

***Operator Overloading*** means giving a special behaviour to an operator for objects.

```python
# Example
class Animal():
    def sound(self):
        print("Animals makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("cat meows")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python

# Real-life Example
class payments():
    def pay(self):
        print("Making payment")

class UPI(payments):
    def pay(self):
        print("Payment through UPI")

class card(payments):
    def pay(self):
        print("Payment through card")

upi = UPI()
Card = card()

upi.pay()
Card.pay()
```

```python
# Method overriding
class animal():
    def sound(self):
        print("Animal makes a sound")

class dog(animal):
    def sound(self):
        print("Dog barks")

Animal = animal()
Dog = dog()

Animal.sound()
Dog.sound() # The child's version overrides the parrent's version
```

```python
# Duck typing
class dog:
    def sound(self):
        print("Bark")

class cat:
    def sound(self):
        print("Meow")

def make_sound(animal):
    animal.sound()

DOG = dog()
CAT = cat()

make_sound(DOG)
make_sound(CAT)
# it doesnot check iit is animal or not it only cares its  has sound() or not.
# Don't care about the type, cares about behaviour.
```

```python
# Operator Overloading
class number():
    def __init__(self,value):
        self.value = value

    def __add__(self,other):
        return self.value + other.value

a = number(10)
b = number(20)

print(a+b)# special Operator, custom behaviouur
```

# 10. Abstraction
***Abstraction*** means hidiing unnecessary implementation details and showing only the essential features to the users.


```python
from abc import ABC, abstractmethod
class Car(ABC):
    @abstractmethod
    def start(self):
        pass

class BMW(Car):
    def start(self):
        print("BMW starts with a button")


class Tesla(Car):
    def start(self):
        print("Tesla starts automatically")

car1 = BMW()
car2 = Tesla()

car1.start()
car2.start()# Every car must have the start function I don't care how u implement it.
```

```python
from abc import ABC, abstractmethod
class animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class dog(animal):
    def sound(self):
        print("Dog barks")

Dog = dog()
Dog.sound()
```

# 11. Encapsulation
***Encapsulation*** is the process of wrapping data and methods together in a class and controlling access to the data.

**Public**    -->Nothing   -->name       -->Directly accessible<br>
**Protected** -->_        -->n_age       -->Accessible, but intended for internal/subclass use<br>
**Private**   -->__        -->__marks    -->Not directly accessible normally

```python
# Example
class Bank():

    def __init__(self):
        self.__balance = 50000

    def deposit(self,amount):
        self.__balance += amount

    def show_balance(self):
        print(self.__balance)

account = Bank()

account.deposit(5000)
account.show_balance()
print(account.show_balance)# show erroe con't directly access
```

# 12. Class and Static method
***class method*** is a method that works with class level data and takes cls as its 1st parameter.<br>
***Static method*** is a method that does not depend on the class or object data.

```python
# class method
class student():

    clg = "alits"

    @classmethod
    def change_clg(cls,new_clg):
        cls.clg = new_clg

print(student.clg)

student.change_clg("jntua")
print(student.clg)
```

```python
# static method
class calculator():

    @staticmethod
    def add(a,b):
        return a+b

print(calculator.add(10,24))
```
